```python
class TokenLimiter:
    """
    A basic implementation to understand token limits.
    For this exercise, we approximate 1 token = 1 word.
    """
    
    def __init__(self, max_tokens):
        # We set the maximum allowed tokens (context window)
        self.max_tokens = max_tokens
        self.data = []
    
    def process(self, input_data):
        """
        Process the input data and enforce the token limit.
        """
        # Requirement 4: Add error handling for edge cases
        if not isinstance(input_data, str):
            return "Error: Input must be a string of text."
            
        if len(input_data.strip()) == 0:
            return "Error: Input text is empty."

        # Split the string into words (approximating tokens)
        tokens = input_data.split()
        token_count = len(tokens)
        
        # Check if the input exceeds the context window
        if token_count > self.max_tokens:
            # Truncate the text to fit the limit
            allowed_tokens = tokens[:self.max_tokens]
            processed_text = " ".join(allowed_tokens)
            return f"[TRUNCATED] {processed_text}..."
        else:
            return f"[OK] {input_data}"


# --- Requirements 2 & 3: Test your implementation and verify output ---

print("--- Test 1: Normal input (Under limit) ---")
# Set limit to 10 tokens
limiter = TokenLimiter(max_tokens=10)
test_1 = "This is a short sentence."
result_1 = limiter.process(test_1)
print(f"Result: {result_1}")
# Expected: [OK] This is a short sentence.

print("\n--- Test 2: Exceeding the context window ---")
test_2 = "This is a very long sentence that definitely exceeds the token limit we set."
result_2 = limiter.process(test_2)
print(f"Result: {result_2}")
# Expected: [TRUNCATED] This is a very long sentence that definitely...

print("\n--- Test 3: Error Handling (Invalid Input) ---")
test_3 = 12345
result_3 = limiter.process(test_3)
print(f"Result: {result_3}")
# Expected: Error: Input must be a string of text.
```

```python
import logging
import matplotlib.pyplot as plt
from typing import Any, Dict, List

# Requirement 5: Write documentation for your implementation
class ProductionSystem:
    """
    A production-ready processor for a Digital Twin.
    This system ingests raw text data, enforces a token limit, 
    logs all actions, handles errors gracefully, and visualizes the results.
    """

    def __init__(self, max_tokens: int = 10):
        # Requirement 1: Combine concepts (Token limits from Part 1)
        self.max_tokens = max_tokens
        self.processed_history = []
        
        # Requirement 3: Include logging capabilities
        # Set up a basic logger to record events
        logging.basicConfig(level=logging.INFO, format='%(levelname)s: %(message)s')
        self.logger = logging.getLogger("DigitalTwin")

    def process(self, input_data: Any) -> Dict[str, Any]:
        """
        Processes incoming data, enforces token limits, and handles errors.
        """
        self.logger.info("Received new data for processing.")
        
        # Requirement 2: Comprehensive error handling
        # Edge case 1: Wrong data type
        if not isinstance(input_data, str):
            self.logger.error(f"Invalid data type received: {type(input_data).__name__}")
            return {"status": "error", "message": "Data must be a string."}
            
        # Edge case 2: Empty string
        if not input_data.strip():
            self.logger.warning("Received empty or whitespace-only string.")
            return {"status": "error", "message": "Data cannot be empty."}

        # Simulate token counting (splitting by spaces)
        tokens = input_data.split()
        token_count = len(tokens)
        
        # Check limits
        if token_count > self.max_tokens:
            self.logger.warning(f"Data exceeds limit ({token_count}/{self.max_tokens}). Truncating.")
            processed_text = " ".join(tokens[:self.max_tokens])
            status = "truncated"
        else:
            self.logger.info(f"Data processed successfully ({token_count}/{self.max_tokens} tokens).")
            processed_text = input_data
            status = "success"

        # Package the result
        result = {
            "status": status,
            "original_length": token_count,
            "processed_text": processed_text
        }
        
        # Save to history for visualization later
        self.processed_history.append(result)
        return result

    # Requirement 4: Create visualization of results
    def visualize_metrics(self):
        """Generates a bar chart showing the token counts of processed data."""
        if not self.processed_history:
            self.logger.warning("No history to visualize yet.")
            return

        # Extract data for the chart
        lengths = [item['original_length'] for item in self.processed_history]
        labels = [f"Run {i+1}" for i in range(len(lengths))]

        plt.figure(figsize=(8, 4))
        bars = plt.bar(labels, lengths, color=['green' if l <= self.max_tokens else 'red' for l in lengths])
        
        # Add a line showing the max token limit
        plt.axhline(y=self.max_tokens, color='black', linestyle='--', label=f'Max Limit ({self.max_tokens})')
        
        plt.title("Digital Twin Data Processing Metrics")
        plt.ylabel("Token Count")
        plt.legend()
        plt.show()


# ==========================================
# Test your production system
# Add comprehensive tests here
# ==========================================

# 1. Initialize the system
twin_system = ProductionSystem(max_tokens=5)

# 2. Run various test cases
print("\n--- RUNNING TESTS ---")
# Test 1: Good data
twin_system.process("Sensor 1 is online.")

# Test 2: Data that is too long (should truncate)
twin_system.process("The pressure in the main valve is currently reading at exactly one hundred PSI.")

# Test 3: Bad data type (should log an error)
twin_system.process(404)

# Test 4: Empty data (should log a warning)
twin_system.process("   ")

# 3. Visualize the results
print("\n--- GENERATING VISUALIZATION ---")
twin_system.visualize_metrics()
```

```python
import re
import logging
import matplotlib.pyplot as plt
from typing import Any, Dict, List, Tuple

# ==========================================
# 1. SETUP LOGGING (Requirement: Handle errors gracefully)
# ==========================================
logging.basicConfig(level=logging.INFO, format='%(levelname)s: %(message)s')
logger = logging.getLogger("TwinSystem")

class DigitalTwinContextManager:
    """
    A complete system to ingest, clean, and chunk digital twin data 
    while respecting context window (token) limits.
    """
    def __init__(self, max_context_tokens: int = 20, overlap_tokens: int = 2):
        self.max_context_tokens = max_context_tokens
        self.overlap_tokens = overlap_tokens
        self.processed_history = []
        logger.info(f"System Initialized. Limit: {max_context_tokens} tokens.")

    # --- Bonus: Optimize for performance (Using Regex) ---
    def clean_data(self, raw_text: str) -> str:
        """Removes noisy log timestamps and extra spaces."""
        # Remove timestamps like [2023-10-25 10:00:00]
        cleaned = re.sub(r'\[.*?\]', '', raw_text)
        # Replace multiple spaces/newlines with a single space
        cleaned = re.sub(r'\s+', ' ', cleaned).strip()
        return cleaned

    def process(self, raw_data: str, source: str = "Twin_Sensor") -> List[Dict[str, Any]]:
        """Processes text into validated chunks."""
        
        # --- Requirement: Handle edge cases gracefully ---
        if not isinstance(raw_data, str):
            logger.error(f"Invalid data type: Expected string, got {type(raw_data).__name__}.")
            return []
        if not raw_data.strip():
            logger.warning("Received empty string. Skipping.")
            return []

        cleaned_text = self.clean_data(raw_data)
        # Approximate tokens by splitting on spaces
        words = cleaned_text.split()
        
        chunks = []
        start_index = 0
        total_words = len(words)

        # Loop to create chunks with overlap
        while start_index < total_words:
            end_index = min(start_index + self.max_context_tokens, total_words)
            chunk_words = words[start_index:end_index]
            
            chunks.append({
                "text": " ".join(chunk_words),
                "source": source,
                "word_count": len(chunk_words)
            })
            
            # If we hit the end, stop looping
            if end_index == total_words:
                break
                
            # Move forward, subtracting the overlap
            start_index += (self.max_context_tokens - self.overlap_tokens)

        # Save for visualization
        self.processed_history.append({
            "source": source, 
            "chunks_created": len(chunks), 
            "total_words": total_words
        })
        
        logger.info(f"Successfully created {len(chunks)} chunks from {source}.")
        return chunks

    # --- Requirement: Create visualizations ---
    def visualize_processing(self):
        """Plots a bar chart showing total words vs chunks created."""
        if not self.processed_history:
            logger.warning("No data to visualize yet.")
            return

        sources = [item['source'] for item in self.processed_history]
        words = [item['total_words'] for item in self.processed_history]
        chunks = [item['chunks_created'] for item in self.processed_history]

        x = range(len(sources))
        
        fig, ax = plt.subplots(figsize=(10, 5))
        
        # Create bars
        ax.bar([i - 0.2 for i in x], words, width=0.4, label='Total Words Ingested', color='skyblue')
        ax.bar([i + 0.2 for i in x], chunks, width=0.4, label='AI Chunks Created', color='salmon')
        
        ax.set_ylabel('Count')
        ax.set_title('Digital Twin Data Processing Metrics')
        ax.set_xticks(x)
        ax.set_xticklabels(sources)
        ax.legend()
        
        plt.show()

# ==========================================
# 2. COMPREHENSIVE TESTING 
# ==========================================
print("\n--- STARTING TESTS ---")

# Initialize with a small limit so we can see the chunking work easily
twin_manager = DigitalTwinContextManager(max_context_tokens=5, overlap_tokens=2)

# Test 1: Messy Log Data (Testing cleaning, chunking, and overlap)
log_1 = "[2023-10-25 10:00] ERROR: Pressure valve is reading 150 PSI. System is overheating. Emergency shutdown initiated."
print("\nRaw Log:", log_1)
chunks_1 = twin_manager.process(log_1, source="Valve_Sensor_1")
for i, c in enumerate(chunks_1):
    print(f"  Chunk {i+1}: '{c['text']}' ({c['word_count']} words)")

# Test 2: Normal Data
twin_manager.process("Temperature is nominal. All systems are operating effectively.", source="Temp_Sensor")

# Test 3: Error Handling (Wrong Type)
print("\nTesting Bad Input...")
twin_manager.process(12345, source="Bad_Input") # Should log an error

# Test 4: Error Handling (Empty String)
twin_manager.process("    ", source="Empty_Input") # Should log a warning

# ==========================================
# 3. VISUALIZATION
# ==========================================
twin_manager.visualize_processing()

# ==========================================
# BONUS: USER-FRIENDLY INTERFACE
# ==========================================
def user_interface():
    print("\n=== DIGITAL TWIN INGESTION TERMINAL ===")
    user_text = "System check: Hydraulic pressure is stable at 120 PSI. Vibration levels are normal."
    print(f"Simulating user input: {user_text}")
    
    result = twin_manager.process(user_text, source="User_Query")
    if result:
        print(f"Success! First chunk ready for AI: {result[0]['text']}")
    else:
        print("Failed to process.")

user_interface()
```
